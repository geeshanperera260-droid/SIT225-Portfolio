```python
from dash import Patch

last_time = None

@app.callback(
    Output("live-graph", "figure"),
    Input("timer", "n_intervals")
)
def smooth_update(n):
    global last_time

    if len(timestamps) == 0:
        return no_update

    time = list(timestamps)[-1]

    # Don't add the same point twice
    if time == last_time:
        return no_update

    last_time = time

    x = list(x_data)[-1]
    y = list(y_data)[-1]
    z = list(z_data)[-1]

    patched = Patch()

    # Add only the new point
    patched["data"][0]["x"].append(time)
    patched["data"][0]["y"].append(x)

    patched["data"][1]["x"].append(time)
    patched["data"][1]["y"].append(y)

    patched["data"][2]["x"].append(time)
    patched["data"][2]["y"].append(z)

    return patched
```

```python
from collections import deque
from datetime import datetime
live_data = deque(maxlen=100)
latest_x = None
latest_y = None
def on_x_sync(client, value):
    global latest_x
    latest_x = value
def on_y_sync(client, value):
    global latest_y
    latest_y = value
def on_z_sync(client, value):
    if latest_x is not None and latest_y is not None:
        live_data.append(
            (
                datetime.now(),
                latest_x,
                latest_y,
                value
            )
        )
client.register("accelerometer_x",value=None,on_write=on_x_sync)
client.register(
    "accelerometer_y",
    value=None,
    on_write=on_y_sync
)

client.register(
    "accelerometer_z",
    value=None,
    on_write=on_z_sync
)

print("Synchronized live buffer ready")
```

```python
@app.callback(
    Output("live-graph", "figure"),
    Input("timer", "n_intervals")
)
def update_sync_graph(n):

    fig = go.Figure()

    if len(live_data) > 0:
        times = [p[0] for p in live_data]
        xs = [p[1] for p in live_data]
        ys = [p[2] for p in live_data]
        zs = [p[3] for p in live_data]

        fig.add_trace(go.Scatter(
            x=times,
            y=xs,
            mode="lines",
            name="X"
        ))

        fig.add_trace(go.Scatter(
            x=times,
            y=ys,
            mode="lines",
            name="Y"
        ))

        fig.add_trace(go.Scatter(
            x=times,
            y=zs,
            mode="lines",
            name="Z"
        ))

    fig.update_layout(
        title="Smooth Live Smartphone Accelerometer",
        xaxis_title="Time",
        yaxis_title="Acceleration"
    )

    return fig
```
